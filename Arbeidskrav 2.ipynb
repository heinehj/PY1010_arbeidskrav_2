```python
# Oppgave 2

antall_elever = int(input("Skriv inn antall elever:"))

# Endret variabelnavn til antall_pizzaer siden det gir mer mening. Divmod funksjon er heltallsdivisjon som gir kvotient og rest. 4 elever pr pizza.
antall_pizzaer, rest = divmod(antall_elever, 4)   

# Om det blir rest, legg til +1. Kan kun kjøpe hele pizzaer.
if rest > 0:
    antall_pizzaer = antall_pizzaer + 1

print("Det trengs", antall_pizzaer, "hele pizzaer")
          
```

```python
# Oppgave 3 

import numpy as np

v_grad = float(input("Skriv inn gradtallet:"))

#Funksjon som regner ut fra grader til radianer.
def fun_v_grad2v_rad(v_grad):
    v_rad = (v_grad * np.pi) / 180
    return v_rad

#Kjører funksjon med "grader-verdi". 
radianer = fun_v_grad2v_rad(v_grad)

print( "Vinkel i radianer:", round(radianer, 3))
```

```python
# Oppgave 5

import numpy as np

# Katet i rettvinklet trekant. "a" har halvsirkel som hatt
a = float(input("Skriv inn lengd på katet a:"))
b = float(input("skriv inn lengd på katet b:"))

# Funksjon som rekner ut areal og omkrets. Husk at radius er halvparten av "a". Bruker pytagoras for å finne hypotenus i omkrets
def fun_areal_omkrets(a, b):
    areal = ((a * b) / 2) + ((np.pi * ((a / 2) ** 2) / 2))
    omkrets = (2 * np.pi * (a / 2)) / 2 + (np.sqrt(a**2 + b**2) + b)
    return areal, omkrets

# Kjører funksjon med katet a og b verdi
areal, omkrets = fun_areal_omkrets(a, b)

print("Areal på figur er:", round(areal, 2))
print("Omkrets på figur er:", round(omkrets, 2))
```

```python
# oppgave 6

import numpy as np
import matplotlib.pyplot as plt

# Lager en array fra -10 til 10 jevnt fordelt på 200 punkt
x = np.linspace(-10 , 10, 200)

# Formel som skal plottes
y = (-x**2) - 5

# Bygger opp grafen
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = -x² - 5")
plt.grid()

# Viser graf på skjerm om program som kjører ikke viser på skjerm autmatisk
plt.show()
```
